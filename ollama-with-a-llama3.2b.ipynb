## Test out the first question

```python
# Quick function to print the answers from the Ollama model
def print_answers(question, desired_model='llama3.2:1b'):
    response = ollama.chat(model=desired_model, messages=[
    {
        'role': 'user',
        'content': question,
    },
    ])

    OllamaResponse=response['message']['content']
    print(OllamaResponse)
```

```python
desired_model='llama3.2:1b'

question = ['What is the best strategy to learn ai?', 
            'What is GenAI',
            'What is the best strategy for building a RAG solution?'
]
```

## Print out the answers for each question

```python
for q in question:
    print_answers(q, desired_model)
    print('---------------------------------------')
```
