## Build a print out function
Two arguements: a ai model that is installed locally & question for the ai.

```python
def print_answers(question, ai_model):
    response = ollama.chat(model=ai_model, messages=[
    {
        'role': 'user',
        'content': question,
        },
        ])
    OllamaResponse=response['message']['content']
    print(OllamaResponse)
```

```python
desired_models = ['llama3.2:1b', 'gemma3', 'mistral']

question = ['What is the best strategy to learn ai?', 
            'What is GenAI',
            'What is the best strategy for building a RAG solution?'
]
```

## Print out the answers for each question

```python
for q in question:
    print_answers(q, 'gemma3')
    print('---------------------------------------')
```
